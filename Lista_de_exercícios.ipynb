```python
print ("Maior entre três números")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))
num3 = float(input("\n Digite o terceiro número:"))

if num1 >= num2 and num1 >= num3:
  print ("\n O primeiro número é o maior!")
elif num2 >= num1 and num2 >= num3:
  print ("\n O segundo número é o maior!")
else:
  print ("\n O terceiro número é o maior!")
```

```python
print ("Número: par ou ímpar")

num = float(input("\n Digite o número:"))

if num % 2 == 0:
  print ("\n O número é par!")

else:
  print ("\n O número é ímpar!")
```

```python
print ("Maior entre três números")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))
num3 = float(input("\n Digite o terceiro número:"))

maior = num1

if num2 > maior:
  maior = num2

if num3 > maior:
  maior = num3

print ("\nEsse é o maior:", maior)

```
