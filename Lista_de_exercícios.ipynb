```python
print("\n 20) Calculadora Simples ")

numero1 = float(input("\nDigite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

print("\nEscolha uma operação:")
print("1 - Soma")
print("2 - Subtração")
print("3 - Multiplicação")
print("4 - Divisão")

operacao = int(input("\nDigite a opção: "))

if operacao == 1:
    resultado = numero1 + numero2
    print("\nResultado:", resultado)

elif operacao == 2:
    resultado = numero1 - numero2
    print("\nResultado:", resultado)

elif operacao == 3:
    resultado = numero1 * numero2
    print("\nResultado:", resultado)

elif operacao == 4:
    if numero2 != 0:
        resultado = numero1 / numero2
        print("\nResultado:", resultado)
    else:
        print("\nNão é possível dividir por zero!")

else:
    print("\nOpção inválida!")
```

```python
print("29) Boletim de notas com média final")

nome = input("\n Nome do(a) aluno(a):")
curso = input ("\n Curso:")
semestre = input ("\n Semestre:")
disciplina = input ("\n Disciplina:")
nota1= float (input("\n Nota Obtida no Primeiro bimestre:"))
nota2= float (input("\n Nota Obtida no Segundo bimestre:"))

media = (nota1+nota2) /2
print ("\n Média final:", media)

if media >100:
    print ("\n LANÇAMENTO INCORRETO!")

if media >=60 and media <=100:
    print ("\n APROVADO!")

elif media <=39:
    print("\n REPROVADO!")

else:
    print ("\n RECUPERAÇÃO!")

print ("\n ----- Relatório -----")
print ("Nome:",nome)
print ("Curso:",curso)
print ("Semestre:",semestre)
print ("Disciplina:",disciplina)
print ("Média:",media)
```
