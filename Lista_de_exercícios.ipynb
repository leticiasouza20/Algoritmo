```python
print ("Adição")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))

soma = num1 + num2

print ("\n A adição é:", soma)
```

```python
print ("Subtração")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))

sub = num1 - num2

print ("\n A subtração é:", sub)
```

```python
print ("Multiplicação")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))

mult = num1 * num2

print ("\n A multiplicação é:", mult)
```

```python
print ("Divisão")

num1 = float(input("\n Digite o primeiro número:"))
num2 = float(input("\n Digite o segundo número:"))

div = num1 / num2

print ("\n A divisão é:", div)
```

```python
print ("Dobro e Triplo")

num = float(input("\n Digite o número:"))

dobro = num*2
triplo = num*3

print ("\n O dobro é:", dobro)
print ("\n O triplo é:", triplo)
```

```python
print ("Antecessor e Sucessor")

num = float(input("\n Digite o número:"))

ant = num - 1
suc = num + 1

print ("\n O antecessor é:", ant)
print ("\n O sucessor é:", suc)
```

```python
print ("Perímetro do Retângulo")

alt = float(input("\n Digite a altura:"))
base = float(input("\n Digite a base:"))

per = 2 * (alt + base)

print ("\n O perímetro do retângulo é:", per)
```

```python
print ("Área do Triângulo")

alt = float(input("\n Digite a altura:"))
base = float(input("\n Digite a base:"))

area = (alt + base) / 2

print ("\n A área do triângulo é:", area)
```
